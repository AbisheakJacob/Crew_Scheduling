```python
from scipy.optimize import linprog

# Define the coefficients for the objective function
c = [-2, -3, -5]

# Define the coefficients matrix for the constraints
A = [[3, 2, 1], [1, 4, 3], [2, 5, 2]]

# Define the right-hand side of the constraints
b = [10, 15, 8]

# Define the bounds for each variable
x_bounds = [(0, None), (0, None), (0, None)]

# Solve the linear programming problem
result = linprog(c, A_ub=A, b_ub=b, bounds=x_bounds, method="highs")

# Extract the optimal dual variables (shadow prices)
dual_variables = result["slack"]

# Print the result and dual variables
print("Optimal Solution:", result["x"])
print(
    "Optimal Value:", -result["fun"]
)  # Note: linprog minimizes the objective function
print("Dual Variables (Shadow Prices):", dual_variables)
```
